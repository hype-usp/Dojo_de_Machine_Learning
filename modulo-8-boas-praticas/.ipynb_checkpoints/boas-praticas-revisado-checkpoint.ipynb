#### Conflitos de versões entre projetos

Durante sua jornada na área, provavelmente você irá estar fazendo vários projetos ao mesmo tempo. Imagine só, dois projetos no trabalho, um projeto na iniciação científica, um projeto no Hype! Muitos projetos diferentes.

Por exemplo, suponha que você vai trabalhar com análise de redes sociais, e vai aplicar um algoritmo de machine learning de uma biblioteca para prever novos relacionamentos na sua rede social. ### Isolando ambientes com máquinas virtuais

Com máquinas virtuais, cada projeto tem seu próprio “universo”, onde suas dependências vivem de forma independente.

Além disso, ambientes virtuais são uma boa prática reconhecida pela comunidade Python, pois ajudam a manter seus projetos organizados, reprodutíveis e fáceis de compartilhar com outras pessoas ou mover para servidores em produção. Com ferramentas como venv ou virtualenv, é possível criar e gerenciar esses ambientes com facilidade.

Vamos voltar ao nosso exemplo, dado que você está dentro do diretório raiz do seu projeto, você pode criar uma máquina virtual com o seguinte comando:

```python
!python -m venv nome_da_maquina_virtual
```

- nome_da_maquina_virtual: nome da máquina virtual que você quer dar para o seu projeto
- Observação: frequentemente o "nome_da_maquina_virtual" que as pessoas colocam é, pasmem: venv

Isso cria uma pasta chamada nome_da_maquina_virtual/ com uma cópia isolada do Python e seus pacotes!

#### Ative o ambiente

Agora você precisa ativar sua maquina virtual (que é como se você estivesse entrando na máquina virtual com seu terminal). 

Se você está no Windows:

```python
venv\\Scripts\\activate
```

Se você está no Linux:

```python
source venv/bin/activate
```

Ótimo! Agora seu terminal está "dentro" da sua máquina virtual! Agora, se você utilizar o comando:

```python
pip list
```

Você deverá ver que você não tem praticamente nada instalado aí, o que significa que essa sua máquina virtual está novinha e pronta para serem instaladas bibliotecas nela!

Eu sei que dificilmente você agora está pensando: "Ok, o ambiente virtual está no terminal, mas como eu vou acessar esse ambiente virtual pelo jupyter lab/jupyter notebook? Será que é só eu abrir o jupyter através desse terminal e o jupyter magicamente estará rodando meu ambiente virtual?"
Estamos na área da Computação, e aqui as coisas devem ser o mais eficiente possíveis, então a resposta não poderia ser outra do que: Não, não é só fazer isso '-'

Obviamente as coisas não poderiam ser tão simples assim, mas não se preocupe e venha comigo que irei te mostrar como acessar o ambiente virtual pelo jupyter.

```python
Dado que seu terminal agora está dentro da máquina virtual, digite:
```

#### Instale o Jupyter e o kernel dentro da máquina virtual

```python
#pip install ipykernel jupyterlab
```

- jupyterlab: instala o Jupyter Lab, caso ainda não tenha.

- ipykernel: permite transformar o ambiente virtual em um kernel reconhecido pelo Jupyter

Mas afinal de contas, o que é um kernel???

No contexto do Jupyter Notebook ou Jupyter Lab, o kernel é o "motor" que executa o código Python. Ele roda no fundo enquanto você escreve no notebook, e é o responsável por processar as células e devolver os resultados.

Cada kernel está ligado a um interpretador Python. Então, quando você cria um ambiente virtual e quer usá-lo no Jupyter, você precisa registrar esse ambiente como um novo kernel, pra que ele apareça como opção lá dentro.

#### Registre esse ambiente como um kernel

```python
python -m ipykernel install --user --name novo_kernel --display-name "Python (meu novo kernel)"
```

--name: nome técnico interno do kernel (sem espaços)

--display-name: nome que aparece no Jupyter

A partir daqui, o Jupyter Lab vai reconhecer seu ambiente virtual como uma opção de kernel.


#### Inicie o Jupyter Lab

```python
jupyter lab
```

Na interface, no canto superior direito provavelmente você verá algo como "Python (ipykernel)", clique nele e você poderá escolher seu kernel indo em:

Python (ipykernel) → Python (meu novo kernel)

#### Remover um kernel

Se quiser apagar esse kernel depois:

```python
jupyter kernelspec uninstall novo_kernel
```

### Como utilizar diferentes versões do python no seu projeto?

✅ Pré-requisito:
Você precisa ter a versão desejada do Python já instalada no seu sistema. Pode ser via:

- Instalador oficial do python.org

💻 1. Crie o ambiente com a versão desejada
Você pode apontar diretamente para o executável Python da versão que quer usar. Por exemplo:

Linux/macOS:

```python
python3.10 -m venv env_projeto
```

Windows (se tiver o Python 3.11 instalado, por exemplo):

```python
py -3.11 -m venv env_projeto
```

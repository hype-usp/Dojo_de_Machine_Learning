## Conflitos de versões entre projetos

Durante sua jornada na área, provavelmente você estará envolvido em vários projetos ao mesmo tempo. Imagine só: dois projetos no trabalho, um projeto na iniciação científica, um projeto no Hype! Muitos projetos diferentes.

É comum que cada um exija versões específicas de bibliotecas. Por exemplo, vamos supor que você vai trabalhar com análise de redes sociais e deseja usar uma biblioteca de machine learning mais antiga. Ela pode exigir uma versão do Pandas que não é compatível com a versão atual do Scikit-learn usada em outro projeto seu.

Imagine que um projeto exige pandas==2.28, e outro precisa do pandas==2.25. Se ambos forem instalados no mesmo sistema, um sobrescreve o outro. Isso pode causar muitos problemas!

Em casos mais extremos, pode até ser necessário usar versões diferentes do Python para cada projeto. Por isso, é importante ter atenção ao ambiente em que está trabalhando!

## Isolando ambientes com máquinas virtuais

Com máquinas virtuais, cada projeto tem seu próprio “universo”, onde suas dependências vivem de forma independente.

Além disso, ambientes virtuais são uma boa prática reconhecida pela comunidade Python, pois ajudam a manter seus projetos organizados, reprodutíveis e fáceis de compartilhar com outras pessoas ou mover para servidores em produção. Com ferramentas como `venv` ou `virtualenv`, é possível criar e gerenciar esses ambientes com facilidade.

Vamos voltar ao nosso exemplo. Suponha que você esteja dentro do diretório raiz do seu projeto. Você pode criar uma máquina virtual com o seguinte comando:

```python
!python -m venv nome_da_maquina_virtual
```

Isso cria uma pasta chamada `nome_da_maquina_virtual/` com uma cópia isolada do Python e seus pacotes!

### Ative o ambiente

Agora você precisa ativar sua máquina virtual (é como se você estivesse “entrando” nela pelo terminal).

Se você está no **Windows**:

```python
venv\Scripts\activate
```

Se você está no **Linux/macOS**:

```python
source venv/bin/activate
```

Ótimo! Agora seu terminal está *dentro* da sua máquina virtual.

Você verá que não há praticamente nada instalado ainda — isso significa que sua máquina virtual está limpa e pronta para receber apenas o que for necessário!

```python
pip list
```

Você pode estar se perguntando: “Ok, o ambiente virtual está ativo no terminal, mas como eu uso esse ambiente no Jupyter Notebook?”

Infelizmente, não é só abrir o Jupyter e esperar que ele use seu novo ambiente automaticamente. Vamos resolver isso!

### Instale o Jupyter e o kernel dentro da máquina virtual

```python
pip install ipykernel jupyterlab
```

Mas afinal, o que é um kernel?

No contexto do Jupyter, o *kernel* é o mecanismo que executa seu código Python. Cada kernel está vinculado a um ambiente (como a máquina virtual que criamos). Por isso, precisamos registrá-lo para que apareça nas opções do Jupyter.

### Registre esse ambiente como um kernel

```python
python -m ipykernel install --user --name meu_kernel --display-name "Python (meu novo kernel)"
```

### Inicie o Jupyter Lab

```python
jupyter lab
```

Na interface, no canto superior direito, clique no nome do kernel (ex: "Python 3") e selecione o novo ambiente: `Python (meu novo kernel)`.
